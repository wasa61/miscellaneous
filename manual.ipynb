```python
import numpy as np
import itertools
import pandas as pd

# 兑换矩阵
matrix = np.array([
    [1,    1.45, 0.52, 0.72],
    [0.7,  1,    0.31, 0.48],
    [1.95, 3.1,  1,    1.49],
    [1.34, 1.98, 0.64, 1]
])

n = 5
start = 3
num_products = matrix.shape[0]

results = []

for steps in range(2, n + 2):
    for path in itertools.product(range(num_products), repeat=steps - 2):
        full_path = [start] + list(path) + [start]
        profit = 1.0
        valid = True
        for i in range(len(full_path) - 1):
            frm, to = full_path[i], full_path[i + 1]
            profit *= matrix[frm][to]
            if profit == 0:
                valid = False
                break
        if valid:
            results.append((full_path, profit))

# 转换为DataFrame输出
df = pd.DataFrame(results, columns=["Path", "Profit"])
df = df.sort_values(by="Profit", ascending=False).reset_index(drop=True)
```
